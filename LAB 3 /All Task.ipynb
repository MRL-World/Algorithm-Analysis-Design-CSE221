```python
 # Task: 2 

import bisect

def merge_sort(arr):
    if len(arr)<=1:
        return arr,0
    mid=len(arr)//2
    
    left,left_inverse=merge_sort(arr[:mid])
    right,right_inverse=merge_sort(arr[mid:])
    merge_array,merge_inverse=merge_and_count(left,right)
    total=left_inverse+right_inverse+merge_inverse
    return merge_array,total

def merge_and_count(left,right):
    count=0
    right_square=[]
    for i in right:
        right_square.append(i*i)
    right_square.sort()
    
    for j in left:
        index=bisect.bisect_left(right_square,j)
        count+=index

    result=[]
    i=j=0
    while i<len(left) and j<len(right):
        if left[i]<=right[j]:
            result.append(left[i])
            i+=1
        else:
            result.append(right[j])
            j+=1
    result.extend(left[i:])
    result.extend(right[j:])
    return result,count

n=int(input())
a=list(map(int,input().split()))
_,receiver=merge_sort(a)
print(receiver)
```

```python
# Task:3

def find_value(a, b, m):
    if b == 0:
        return 1

    if b % 2 == 0:
        half = find_value(a, b // 2, m)
        result = (half * half) % m
        return result
    else:
        prev_power = find_value(a, b - 1, m) 
        after_mul = a * prev_power          
        result = after_mul % m          
        return result                     
a, b = map(int, input().split())
print(find_value(a, b, 107))
```

```python
# Task :4

MOD = 10**9 + 7
def multiply_matrix(A, B):
    return [
        
        
        [(A[0][0] * B[0][0] + A[0][1] * B[1][0]) % MOD, (A[0][0] * B[0][1] + A[0][1] * B[1][1]) % MOD],
        [(A[1][0] * B[0][0] + A[1][1] * B[1][0]) % MOD, (A[1][0] * B[0][1] + A[1][1] * B[1][1]) % MOD]
    ]

def power_matrix(A, X):
    result = [[1, 0], [0, 1]]

    while X > 0:
        if X % 2 == 1:
            result = multiply_matrix(result, A)
        A = multiply_matrix(A, A)
        X //= 2
    
    return result

T = int(input())

for i in range(T):
    a11, a12, a21, a22 = map(int, input().split())
    X = int(input())
    A = [[a11, a12], [a21, a22]]
    result = power_matrix(A, X)
    print(result[0][0], result[0][1])
    print(result[1][0], result[1][1])
```

```python
# Task: 5
def modulas_power(a, b, m):
    res = 1
    a = a % m
    while b > 0:
        if b % 2:
            res = (res * a) % m
        a = (a * a) % m
        b //= 2
    return res

def geomaetric_sum(a, n, m):
    if n == 1:
        return a % m
    if n % 2 == 0:
        mid = geomaetric_sum(a, n//2, m)
        ap = modulas_power(a, n//2, m)
        return (mid * (1 + ap)) % m
    else:
        before = geomaetric_sum(a, n-1, m)
        term = modulas_power(a, n, m)
        return (before + term) % m

T = int(input())
for i in range(T):
    a, n, m = map(int, input().split())
    if a == 1:
        print(n % m)
    else:
        print(geomaetric_sum(a, n, m))
```

```python
# Task : 6

def build_tree(arr, left, right, result):
    if left > right:
        return
    mid = (left + right) // 2  
    result.append(arr[mid])  
    build_tree(arr, left, mid - 1, result)  
    build_tree(arr, mid + 1, right, result)  


n = int(input())  
arr = list(map(int, input().split()))  
result = []
build_tree(arr, 0, n - 1, result)
print(*result)  
```

```python
# Task:7

n = int(input())  
in_order = list(map(int, input().split()))  
pre_order = list(map(int, input().split()))  

def post_order(in_order, pre_order):
    if not in_order:
        return []
    
    root = pre_order[0]  
    
    root_index = in_order.index(root)  

    left_subtree = post_order(in_order[:root_index], pre_order[1:root_index+1])
    right_subtree = post_order(in_order[root_index+1:], pre_order[root_index+1:])
    
    return left_subtree + right_subtree + [root]  


print(*post_order(in_order, pre_order))
```

```python
# Task : 8 

n = int(input())
in_order = list(map(int, input().split()))
post_order = list(map(int, input().split()))

def pre_order(in_order, post_order):
    if not in_order:
        return []
    
    root = post_order[-1]
    root_index = in_order.index(root)
    
    left = pre_order(in_order[:root_index], post_order[:root_index])
    right = pre_order(in_order[root_index+1:], post_order[root_index:-1])
    
    return [root] + left + right

print(" ".join(map(str, pre_order(in_order, post_order))))
```
