```python
# Task: A (Topological Sort for Course Scheduling)
from queue import Queue

def solve():
    n, m = map(int, input().split())
    
    graph = [[] for _ in range(n+1)]

    indegree = [0] * (n+1)
    
    for _ in range(m):
        a, b = map(int, input().split())
        graph[a].append(b)      
        indegree[b] += 1      
    q = Queue()
    for i in range(1, n+1):
        if indegree[i] == 0:
            q.put(i)
    
    result = []  
    while not q.empty():
        course = q.get() 
        result.append(course)
        for j in graph[course]:
            indegree[j] -= 1
            if indegree[j] == 0:
                q.put(j)
    
    if len(result) == n:
        print(" ".join(map(str, result)))
    else:
        print(-1)
        
if __name__ == "__main__":
    solve()
```

```python
# Task: B (Bipartite Graph Coloring)

from collections import deque

def solve():
    n, m = map(int, input().split())
    
    graph = [[] for _ in range(n+1)]
    for _ in range(m):
        u, v = map(int, input().split())
        graph[u].append(v)
        graph[v].append(u)
    
    color = [-1] * (n+1)  
    result = 0

    for i in range(1, n+1):
        if color[i] == -1:
            queue = deque()
            queue.append(i)
            color[i] = 0
            count = [1, 0]  

            while queue:
                u = queue.popleft()
                for v in graph[u]:
                    if color[v] == -1:
                        color[v] = 1 - color[u]
                        count[color[v]] += 1
                        queue.append(v)

            result += max(count)  # এই component এ max robots/humans

    print(result)
if __name__ == "__main__":
    solve()
```

```python
# Task: C (Knight's Shortest Path on a Chessboard)

from collections import deque

N = int(input())
x1, y1, x2, y2 = map(int, input().split())

moves = [(2, 1), (2, -1), (-2, 1), (-2, -1),
         (1, 2), (1, -2), (-1, 2), (-1, -2)]

dist = [[-1] * (N + 1) for _ in range(N + 1)]
dist[x1][y1] = 0

dq = deque([(x1, y1)])

while dq:
    x, y = dq.popleft()
    if (x, y) == (x2, y2):
        print(dist[x][y])
        break

    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        if 1 <= nx <= N and 1 <= ny <= N and dist[nx][ny] == -1:
            dist[nx][ny] = dist[x][y] + 1
            dq.append((nx, ny))
            
if dist[x2][y2] == -1:
    print(-1)
```

```python
# Task: D (Tree Diameter)

from collections import defaultdict, deque

n = int(input())
tree = defaultdict(list)
for _ in range(n-1):
    a, b = map(int, input().split())
    tree[a].append(b)
    tree[b].append(a)

def bfs_traversal(tree, start_node):
    distance = defaultdict(int)
    seen = set()
    distance[start_node] = 0
    seen.add(start_node)
    q = deque([start_node])
    
    last_node = start_node
    longest = 0

    while q:
        node = q.popleft()
        for i in tree[node]:
            if i not in seen:
                seen.add(i)
                distance[i] = distance[node] + 1
                q.append(i)
                if distance[i] > longest:
                    longest = distance[i]
                    last_node = i

    return last_node, longest

node1, _ = bfs_traversal(tree, 1)

node2, diameter = bfs_traversal(tree, node1)

print(diameter)
print(node1, node2)
```

```python
# Task: E (Alien Dictionary)

from collections import defaultdict, deque
import heapq

n = int(input())
words = [input().strip() for _ in range(n)]

edges = defaultdict(list)
indegree = defaultdict(int)
letters = set()

for k in range(n - 1):
    w1, w2 = words[k], words[k + 1]

    if len(w1) > len(w2) and w1.startswith(w2):
        print(-1)
        exit()

    for idx in range(min(len(w1), len(w2))):
        if w1[idx] != w2[idx]:
            edges[w1[idx]].append(w2[idx])
            indegree[w2[idx]] += 1
            break

for w in words:
    for ch in w:
        letters.add(ch)


pq = []
for ch in letters:
    if indegree[ch] == 0:
        heapq.heappush(pq, ch)

order = []

while pq:
    node = heapq.heappop(pq)
    order.append(node)

    for nxt in edges[node]:
        indegree[nxt] -= 1
        if indegree[nxt] == 0:
            heapq.heappush(pq, nxt)

if len(order) == len(letters):
    print("".join(order))
else:
    print(-1)
```

```python
# Task: F 

from collections import deque, defaultdict

N, M, S, Q = map(int, input().split())

graph = defaultdict(list)

for _ in range(M):
    u, v = map(int, input().split())
    graph[u].append(v)
    graph[v].append(u)

sources = list(map(int, input().split()))

destinations = list(map(int, input().split()))

dist = [-1] * (N + 1)

q = deque()
for s in sources:
    dist[s] = 0
    q.append(s)

while q:
    node = q.popleft()
    for i in graph[node]:
        if dist[i] == -1:   # not visited yet
            dist[i] = dist[node] + 1
            q.append(i)
ans = []
for d in destinations:
    ans.append(str(dist[d]))
print(" ".join(ans))
```
