```python
# Task:A  (program to check whether a number is even or odd)

T=int(input())
for i in range(T):
    user=int(input())
    if user%2==0:
        print(f"{user} is an Even number.")
    else:
        print(f"{user} is an Odd number.")
```

```python
# Task:B 

T=int(input())
for i in range(T):
    user=input().split()
    first=int(user[1])
    second=int(user[3])
    sign=user[2]
    if sign=="+":
        result=float(first+second)
    elif sign=="/":
        result=float(first/second)
    elif sign=="*":
        result=float(first*second)
    elif sign=="-":
        result=float(first-second)
    print(result)
```

```python
# Task:E
def can_sort(arr):
    sorted_arr = sorted(arr)
    if arr == sorted_arr:
        return "YES"
    if len(arr) < 3:
        return "NO"

    sorted_evens = [sorted_arr[i] for i in range(0, len(arr), 2)]
    sorted_odds = [sorted_arr[i] for i in range(1, len(arr), 2)]

    og_evens = [arr[i] for i in range(0, len(arr), 2)]
    og_odds = [arr[i] for i in range(1, len(arr), 2)]



    og_evens = sorted(og_evens)
    og_odds = sorted(og_odds)

    if og_evens == sorted_evens and og_odds == sorted_odds:
        return "YES"
    else:
        return "NO"

N = int(input())
arr = list(map(int, input().split()))
print(can_sort(arr))
```

```python
# Task:G

N = int(input())
arr1 = list(map(int, input().split()))
arr2 = list(map(int, input().split()))
count = 0

student = []
for i in range(len(arr1)):
  student.append((arr1[i],arr2[i]))


for i in range(len(student)-1):
  max_idx = i
  for j in range(i + 1, len(student)):
      if student[j][1] > student[max_idx][1]:
        max_idx = j
      elif student[j][1] == student[max_idx][1] and student[j][0] < student[max_idx][0]:
        max_idx = j
  if i != max_idx:
    student[i], student[max_idx] = student[max_idx], student[i]
    count += 1

print("Minimum swaps:",count)

for i in range(len(student)):
  print(f"ID: {student[i][0]} Mark: {student[i][1]}")
```
