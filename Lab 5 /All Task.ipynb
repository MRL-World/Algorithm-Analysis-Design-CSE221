```python
# Task A

n, m = map(int, input().split())

adj = [[] for _ in range(n + 1)]

for l in range(m):
    u, v = map(int, input().split())
    adj[u].append(v)
    adj[v].append(u)

def bfs(start):
    visited = [False] * (n + 1)
    order = []
    q = [start]
    head = 0
    visited[start] = True

    while head < len(q):
        node = q[head]
        head += 1
        order.append(node)

        for i in sorted(adj[node]):
            if not visited[i]:
                visited[i] = True
                q.append(i)

    return order

result = bfs(1)
print(" ".join(map(str, result)))
```

```python
# Task : B

n, m = map(int, input().split())
u = list(map(int, input().split()))
v = list(map(int, input().split()))

graph = [[] for l in range(n + 1)]
for i in range(m):
    graph[u[i]].append(v[i])
    graph[v[i]].append(u[i])
    
for i in range(1, n + 1):
    graph[i].sort()

visited = [False] * (n + 1)
order = []

my_stack = [1]
while my_stack:
    node = my_stack.pop()
    if visited[node]:
        continue
    visited[node] = True
    order.append(str(node))
    
    for k in reversed(graph[node]):
        if not visited[k]:
            my_stack.append(k)

print(" ".join(order))
```

```python
# Task : C

n, m, s, d = map(int, input().split())
u = list(map(int, input().split()))
v = list(map(int, input().split()))

graph = [[] for l in range(n + 1)]
for i in range(m):
    graph[u[i]].append(v[i])
    graph[v[i]].append(u[i])

for i in range(1, n + 1):
    graph[i].sort()

distance = [-1] * (n + 1)
parent = [-1] * (n + 1)

my_queue = [s]
front = 0  
distance[s] = 0

while front < len(my_queue):
    node = my_queue[front]
    front += 1
    for i in graph[node]:
        if distance[i] == -1:
            distance[i] = distance[node] + 1
            parent[i] = node
            my_queue.append(i)

if distance[d] == -1:
    print(-1)
else:
    path = []
    cur = d
    while cur != -1:
        path.append(cur)
        cur = parent[cur]
    path.reverse()

    print(distance[d])
    print(" ".join(map(str, path)))
```

```python
# Task : D

def bfs(n, adj, start, end):
    distance = [-1] * (n + 1)
    parent = [-1] * (n + 1)
    my_queue = [start]
    head = 0
    distance[start] = 0

    while head < len(my_queue):
        node = my_queue[head]
        head += 1
        for i in adj[node]:
            if distance[i] == -1:
                distance[i] = distance[node] + 1
                parent[i] = node
                my_queue.append(i)
                if i == end:
                    return distance, parent
    return distance, parent


n, m, s, d, k = map(int, input().split())
adj = [[] for _ in range(n + 1)]
for _ in range(m): 
    u, v = map(int, input().split())
    adj[u].append(v)

distance1, parent1 = bfs(n, adj, s, k)
if distance1[k] == -1:
    print(-1)
    quit()

distance2, parent2 = bfs(n, adj, k, d)
if distance2[d] == -1:
    print(-1)
    quit()

path1 = []
cur = k
while cur != -1:
    path1.append(cur)
    cur = parent1[cur]
path1.reverse()

path2 = []
cur = d
while cur != -1:
    path2.append(cur)
    cur = parent2[cur]
path2.reverse()

full_path = path1 + path2[1:]

print(len(full_path) - 1)
print(" ".join(map(str, full_path)))
```

```python
# Task : E

n, r = map(int, input().split())

graph = [[] for _ in range(n + 1)]
for _ in range(n - 1):
    a, b = map(int, input().split())
    graph[a].append(b)
    graph[b].append(a)

subtree = [0] * (n + 1)

my_stack = [(r, -1, 0)] 

while my_stack:
    node, parent, state = my_stack.pop()
    if state == 0:
        my_stack.append((node, parent, 1)) 
        for child in graph[node]:
            if child != parent:
                my_stack.append((child, node, 0))
    else:
        size = 1
        for child in graph[node]:
            if child != parent:
                size += subtree[child]
        subtree[node] = size

q = int(input())
for _ in range(q):
    x = int(input())
    print(subtree[x])
```

```python
# Task : F

n, m = map(int, input().split())

graph = [[] for _ in range(n + 1)]
for _ in range(m):
    u, v = map(int, input().split())
    graph[u].append(v)

visited = [False] * (n + 1)
my_stack = [False] * (n + 1)
found_cycle = False

for i in range(1, n + 1):
    if visited[i]:
        continue

    stack = [(i, 0)] 
    stack_path = []

    while stack:
        node, index = stack.pop()
        if index == 0: 
            visited[node] = True
            my_stack[node] = True
            stack_path.append(node)

        if index < len(graph[node]):
            neighbor = graph[node][index]
            stack.append((node, index + 1))
            if not visited[neighbor]:
                stack.append((neighbor, 0))
            elif my_stack[neighbor]:
                found_cycle = True
                break
        else:
            my_stack[node] = False
            stack_path.pop()

    if found_cycle:
        break

if found_cycle:
    print("YES")
else:
    print("NO")
```

```python
# Task: G

R, H = map(int, input().split())
grid = [list(input().strip()) for _ in range(R)]
visited = [[False] * H for _ in range(R)]

directions = [(-1,0), (1,0), (0,-1), (0,1)]

def bfs(sr, sc):
    q = [(sr, sc)]
    head = 0
    visited[sr][sc] = True
    diamonds = 0
    
    while head < len(q):
        r, c = q[head]
        head += 1
        
        if grid[r][c] == 'D':
            diamonds += 1
        
        for dr, dc in directions:
            nr, nc = r + dr, c + dc
            if 0 <= nr < R and 0 <= nc < H:
                if not visited[nr][nc] and grid[nr][nc] != '#':
                    visited[nr][nc] = True
                    q.append((nr, nc))
    return diamonds

max_diamonds = 0

for i in range(R):
    for j in range(H):
        if not visited[i][j] and grid[i][j] != '#':
            count = bfs(i, j)
            if count > max_diamonds:
                max_diamonds = count

print(max_diamonds)
```
