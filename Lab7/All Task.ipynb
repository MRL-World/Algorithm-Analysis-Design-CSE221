```python
# Task: A - Shortest Path diajkstra

import heapq

def sortest_path(n, adj, source, dest):
    initial_value = 10**18
    dist = [initial_value] * (n + 1)
    parent = [-1] * (n + 1)

    dist[source] = 0
    pq = []
    heapq.heappush(pq, (0, source))   

    while pq:
        d, u = heapq.heappop(pq)
        if d > dist[u]:
            continue
        for v, w in adj[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    if dist[dest] == initial_value:
        return -1, []

    path = []
    cur = dest
    while cur != -1:
        path.append(cur)
        cur = parent[cur]
    path.reverse()
    return dist[dest], path

N, M, S, D = map(int, input().split())
u = list(map(int, input().split()))
v = list(map(int, input().split()))
w = list(map(int, input().split()))

adj = [[] for _ in range(N + 1)]
for i in range(M):
    adj[u[i]].append((v[i], w[i]))

dist, path = sortest_path(N, adj, S, D)

if dist == -1:
    print(-1)
else:
    print(dist)
    print(*path)
```

```python
# Task : B - Meeting Point 

import heapq

def dijkstra(n, adj, source):
    initial_value = 10**18
    dist = [initial_value] * (n + 1)
    dist[source] = 0
    pq = [(0, source)]
    while pq:
        d, u = heapq.heappop(pq)
        if d > dist[u]:
            continue
        for v, w in adj[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                heapq.heappush(pq, (dist[v], v))
    return dist


N, M, S, T = map(int, input().split())
adj = [[] for _ in range(N + 1)]

for i in range(M):
    u, v, w = map(int, input().split())
    adj[u].append((v, w))

distA = dijkstra(N, adj, S)

distB = dijkstra(N, adj, T)

initial_value = 10**18
best_time = initial_value
best_node = -1

for i in range(1, N + 1):
    if distA[i] < initial_value and distB[i] < initial_value:
        meet_time = max(distA[i], distB[i])
        if meet_time < best_time or (meet_time == best_time and i < best_node):
            best_time = meet_time
            best_node = i

if best_node == -1:
    print(-1)
else:
    print(best_time, best_node)
```

```python
# Task: C  - Minimize the Danger

import heapq

N, M = map(int, input().split())
adj = [[] for _ in range(N + 1)]

for i in range(M):
    u, v, w = map(int, input().split())
    adj[u].append((v, w))
    adj[v].append((u, w))  

inital_value = 10**18
danger = [inital_value] * (N + 1)
danger[1] = 0

pq = [(0, 1)]  

while pq:
    d, u = heapq.heappop(pq)
    if d > danger[u]:
        continue
    for v, w in adj[u]:
        new_d = max(d, w)
        if new_d < danger[v]:
            danger[v] = new_d
            heapq.heappush(pq, (new_d, v))
            
for i in range(1, N + 1):
    if danger[i] == inital_value:
        print(-1, end=' ')
    else:
        print(danger[i], end=' ')
```

```python
# Task: D  - Node Weight Shortest Path

import heapq

N, M, S, D = map(int, input().split())
node_weight = [0] + list(map(int, input().split()))  

adj = [[] for _ in range(N + 1)]
for _ in range(M):
    u, v = map(int, input().split())
    adj[u].append(v)


initial_value = 10**18
dist = [initial_value] * (N + 1)
dist[S] = node_weight[S]

pq = [(dist[S], S)]  

while pq:
    cost, u = heapq.heappop(pq)
    if cost > dist[u]:
        continue
    for v in adj[u]:
        new_cost = cost + node_weight[v]
        if new_cost < dist[v]:
            dist[v] = new_cost
            heapq.heappush(pq, (dist[v], v))

if dist[D] == initial_value:
    print(-1)
else:
    print(dist[D])
```

```python
# Task : E Parity Edges

import heapq

N, M = map(int, input().split())
u = list(map(int, input().split()))
v = list(map(int, input().split()))
w = list(map(int, input().split()))

adj = [[] for _ in range(N + 1)]
for i in range(M):
    adj[u[i]].append((v[i], w[i]))


initial_value = 10**18
dist = [[initial_value, initial_value] for _ in range(N + 1)]  

pq = []

for next_node, weight in adj[1]:
    parity = weight % 2
    dist[next_node][parity] = weight
    heapq.heappush(pq, (weight, next_node, parity))

while pq:
    d, u, last_parity = heapq.heappop(pq)
    if d > dist[u][last_parity]:
        continue
    for v, w in adj[u]:
        parity = w % 2
        if parity == last_parity:
            continue  
        new_dist = d + w
        if new_dist < dist[v][parity]:
            dist[v][parity] = new_dist
            heapq.heappush(pq, (new_dist, v, parity))


res = min(dist[N])
if res == initial_value:
    print(-1)
else:
    print(res)
```

```python
# Task : F Shortest Path Revisited

import heapq

N, M, S, D = map(int, input().split())
adj = [[] for _ in range(N + 1)]

for _ in range(M):
    u, v, w = map(int, input().split())
    adj[u].append((v, w))
    adj[v].append((u, w)) 


initial_value = 10**18
first = [initial_value] * (N + 1)
second = [initial_value] * (N + 1)

pq = []
first[S] = 0
heapq.heappush(pq, (0, S))

while pq:
    d, u = heapq.heappop(pq)
    for v, w in adj[u]:
        nd = d + w
        if nd < first[v]:
            second[v] = first[v]
            first[v] = nd
            heapq.heappush(pq, (first[v], v))
        elif first[v] < nd < second[v]:
            second[v] = nd
            heapq.heappush(pq, (second[v], v))

if second[D] == initial_value:
    print(-1)
else:
    print(second[D])
```
